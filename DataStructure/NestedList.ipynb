```python
matrix_A = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matrix_B = [
    [9, 8, 7],
    [6, 5, 4],
    [3, 2, 1]
]
print("Matrix A:", matrix_A)
print("Matrix B:", matrix_B)
```

```python
matrix_C = []
for i in range(3):
    for j in range(3):
        matrix_C.append(matrix_A[i][j] - matrix_B[i][j])
print("Matrix C (A - B) flattened:", matrix_C)
```
